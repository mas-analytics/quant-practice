```python
def backtest_ma(ticker, window):

    df = yf.download(ticker, period="5y")

    close = df["Close"][ticker]

    returns = close.pct_change()

    ma = close.rolling(window).mean()

    signal = close > ma

    strategy_returns = (
        signal.shift(1)
        .fillna(False)
        .astype(int)
        * returns
    )

    cumulative_returns = (
        1 + strategy_returns
    ).cumprod()

    return cumulative_returns
```
