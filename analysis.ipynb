```python
def backtest_ma(ticker, window):

    df = yf.download(ticker, period="5y")

    close = df["Close"][ticker]

    returns = close.pct_change()

    ma = close.rolling(window).mean()

    signal = close > ma

    strategy_returns = (
        signal.shift(1)
        .fillna(False)
        .astype(int)
        * returns
    )

    cumulative_returns = (
        1 + strategy_returns
    ).cumprod()

    return cumulative_returns
```

# Quant Research Notebook

このNotebookは、MA・Momentum・Buy & Hold戦略の比較と、
MAパラメータの頑健性検証を行う。

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def backtest_buy_hold(ticker):
    df = yf.download(ticker, period="5y")
    close = df["Close"][ticker]
    returns = close.pct_change()
    cumulative_returns = (1 + returns).cumprod()
    rolling_max = cumulative_returns.cummax()
    drawdown = cumulative_returns / rolling_max - 1
    max_drawdown = drawdown.min()
    annual_sharpe = (
        returns.mean()
        / returns.std()
        * np.sqrt(252)
    )
    final_return = cumulative_returns.iloc[-1]
    return {"Ticker":ticker,
            "Strategy":"Buy & Hold",
            "Window":np.nan,
            "Threshold":np.nan,
            "Final Return":final_return,
            "Max Drawdown":max_drawdown,
            "Sharpe":annual_sharpe,
    }

def backtest_ma(ticker, window):
    df = yf.download(ticker, period="5y")
    close = df["Close"][ticker]
    returns = close.pct_change()
    ma = close.rolling(window).mean()
    signal = close > ma
    strategy_returns = (
        signal.shift(1)
        .fillna(False)
        .astype(int)
        * returns
    )
    cumulative_returns = (
        1 + strategy_returns
    ).cumprod()
    rolling_max = cumulative_returns.cummax()
    drawdown = (
        cumulative_returns
        / rolling_max
        - 1
    )
    max_drawdown = drawdown.min()
    annual_sharpe = (
        strategy_returns.mean()
        / strategy_returns.std()
        * np.sqrt(252)
    )
    final_return = cumulative_returns.iloc[-1]
    return {"Ticker":ticker,
            "Strategy":"Moving Average",
            "Window":window,
            "Threshold":np.nan,
            "Final Return":final_return,
            "Max Drawdown":max_drawdown,
            "Sharpe":annual_sharpe
    }

def backtest_momentum(ticker, window, threshold):
    df = yf.download(ticker, period="5y")
    close = df["Close"][ticker]
    returns=close.pct_change()
    momentum=close/close.shift(window)-1
    signal=momentum>threshold
    strategy_returns=returns*signal.shift(1)
    cumulative_returns=(1+strategy_returns).cumprod()
    rolling_max=cumulative_returns.cummax()
    drawdown=cumulative_returns/rolling_max-1
    max_drawdown=drawdown.min()
    annual_sharpe=strategy_returns.mean()/strategy_returns.std()*np.sqrt(252)
    final_return=cumulative_returns.iloc[-1]
    return {"Ticker":ticker,
            "Strategy":"Momentum",
            "Window":window,
            "Threshold":threshold,
            "Final Return":final_return,
            "Max Drawdown":max_drawdown,
            "Sharpe":annual_sharpe
    }
```
